```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder

file_path = r"C:\Users\ashun\Downloads\flight_data_2024_final_cleaned.csv"
df = pd.read_csv(file_path, low_memory=False)

# Create a date-based feature
df["fl_date"] = pd.to_datetime(df["fl_date"], errors="coerce")
df["day_of_year"] = df["fl_date"].dt.dayofyear

# Select numerical features for transformation
features = ["dep_delay", "arr_delay", "taxi_out",
            "taxi_in", "air_time", "distance"]

feature_df = df[features].dropna().sample(
    n=50000, random_state=42
)

# Standardization
standard_scaler = StandardScaler()
standardized = standard_scaler.fit_transform(feature_df)

# Min-Max scaling
minmax_scaler = MinMaxScaler()
normalized = minmax_scaler.fit_transform(feature_df)

print("Features selected:", features)
print("Sample shape:", feature_df.shape)
print("\nStandardized means:")
print(pd.Series(standardized.mean(axis=0),
                index=features).round(3))

print("\nMin-Max scaled ranges:")
print(pd.DataFrame(normalized, columns=features).agg(["min", "max"]).round(2))
```

- Applied StandardScaler to standardize features.
