# Exploratory Data Analysis – Airline Flight Delay Dataset

## Week 1 EDA

This notebook contains the initial exploration of the 2024 airline flight delay dataset. The aim is to understand the dataset, identify important variables, and find initial patterns that can be useful for the flight delay dashboard.

## 1. Dataset Overview

The dataset contains flight information for the year 2024, including airline details, departure and arrival information, flight delays, cancellations, and different causes of delays.

The cleaned dataset contains 96,711 rows and 35 columns.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df['dep_delay'], bins=30)
plt.xlabel('Departure Delay (minutes)')
plt.ylabel('Number of Flights')
plt.title('Distribution of Departure Delays')
plt.show()
```

```python
avg_delay = df.groupby('op_unique_carrier')['dep_delay'].mean().sort_values(ascending=False)

plt.figure(figsize=(9,5))
avg_delay.plot(kind='bar')
plt.xlabel('Airline')
plt.ylabel('Average Departure Delay (minutes)')
plt.title('Average Departure Delay by Airline')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
cancel_counts = df['cancelled'].value_counts()

plt.figure(figsize=(6,4))
cancel_counts.plot(kind='bar')
plt.xlabel('Cancelled (0 = No, 1 = Yes)')
plt.ylabel('Number of Flights')
plt.title('Flight Cancellation Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
day_delay = df.groupby('day_of_week')['dep_delay'].mean()

plt.figure(figsize=(8,5))
day_delay.plot(kind='bar')
plt.xlabel('Day of Week')
plt.ylabel('Average Departure Delay (minutes)')
plt.title('Average Departure Delay by Day of Week')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['dep_delay'], df['arr_delay'], alpha=0.3)
plt.xlabel('Departure Delay (minutes)')
plt.ylabel('Arrival Delay (minutes)')
plt.title('Departure Delay vs Arrival Delay')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['arr_delay'], bins=30)
plt.xlabel('Arrival Delay (minutes)')
plt.ylabel('Number of Flights')
plt.title('Distribution of Arrival Delays')
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8,5))
plt.hist(df['arr_delay'], bins=30)
plt.xlabel('Arrival Delay (minutes)')
plt.ylabel('Number of Flights')
plt.title('Distribution of Arrival Delays')
plt.tight_layout()
plt.show()
```

## Key Findings

- The dataset contains 96,711 flight records and 35 columns.
- Only one missing value was found in the `fl_date` column.
- WN has the highest number of flights in the dataset.
- B6 has the highest average departure delay, while YX has the lowest.
- Day 6 has the highest average departure delay among the recorded days.
- Departure delay and arrival delay show a strong positive correlation of 0.73.
- Only 324 flights were cancelled, compared with 96,387 non-cancelled flights.
- The departure and arrival delay distributions show variation in flight timing performance.
