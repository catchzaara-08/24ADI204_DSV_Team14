# Exploratory Data Analysis – Airline Flight Delay Dataset

## Week 1 EDA

This notebook contains the initial exploration of the 2024 airline flight delay dataset. The aim is to understand the dataset, identify important variables, and find initial patterns that can be useful for the flight delay dashboard.

## 1. Dataset Overview

The dataset contains flight information for the year 2024, including airline details, departure and arrival information, flight delays, cancellations, and different causes of delays.

The sample used for initial exploration contains **10,000 rows and 35 columns**.

```python
import pandas as pd
df = pd.read_csv("../data/raw/flight_data_2024_sample.csv")
```

## 2. Airline Distribution

The distribution of flights across different airlines was examined to understand how the sample is represented across carriers.

```python
df['op_unique_carrier'].value_counts()
```

### Observation

WN has the highest number of flights in the sample, followed by DL, AA, OO, and UA. The number of flights varies across the different carriers.

## 3. Cancellation Analysis

The cancellation status of flights was examined to understand the proportion of cancelled and non-cancelled flights in the sample.

```python
df['cancelled'].value_counts()
```

### Observation

Out of 10,000 flights in the sample, 9,878 were not cancelled and 122 were cancelled. This shows that cancellations form a small portion of the sample.

```python
(df['dep_delay'] > 0).value_counts()
```

## 4. Departure Delay Analysis

The sample was examined to identify the number of flights with and without recorded departure delays.

```python
df.loc[df['dep_delay'].notna(), 'dep_delay'].gt(0).value_counts()
```

### Observation

Among the 9,884 flights with recorded departure-delay values, 3,663 flights had a positive departure delay, while 6,221 flights did not have a positive departure delay. This indicates that a considerable portion of the recorded flights experienced departure delays.

```python
df[df['dep_delay'] > 0].groupby('op_unique_carrier').size().sort_values(ascending=False)
```

## 5. Airline-wise Delay Rate

The departure delay rate was compared across airlines to identify differences in delay frequency between carriers.

```python
df.groupby('op_unique_carrier')['dep_delay'].apply(lambda x: (x > 0).mean() * 100).sort_values(ascending=False)
```

### Observation

In the sample, WN has the highest departure-delay rate at about 47.3%, while YX has the lowest at about 17.5%. The results show noticeable differences in delay rates across airlines.

```python
df['dep_delay'].describe()
```

## 6. Monthly Flight Distribution

The number of flights in the sample was examined across the 12 months of 2024 to understand how the sample is distributed over time.

```python
df['month'].value_counts().sort_index()
```

### Observation

The sample contains flights from all 12 months of 2024. The number of flights varies across months, with October having the highest representation and February having the lowest.
