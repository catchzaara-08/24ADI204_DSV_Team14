```python
## Bivariate EDA

Bivariate analysis is used to study the relationship between two variables and identify patterns or associations in the flight dataset.
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(df["dep_delay"], df["arr_delay"], alpha=0.3)
plt.xlabel("Departure Delay (minutes)")
plt.ylabel("Arrival Delay (minutes)")
plt.title("Departure Delay vs Arrival Delay")
plt.show()
```

```python
correlation = df["dep_delay"].corr(df["arr_delay"])
print("Correlation between departure delay and arrival delay:", round(correlation, 3))
```

### Observation

There is a strong positive correlation (0.734) between departure delay and arrival delay. This indicates that flights with higher departure delays generally tend to have higher arrival delays as well. However, the scatter plot shows considerable variation, so the relationship is not perfectly linear.

```python
avg_arr_delay = (
    df.groupby("op_unique_carrier")["arr_delay"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))
avg_arr_delay.plot(kind="bar")
plt.xlabel("Airline")
plt.ylabel("Average Arrival Delay (minutes)")
plt.title("Average Arrival Delay by Airline")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print(avg_arr_delay.round(2))
```

### Observation

Average arrival delay varies considerably across airlines. HA has the highest average arrival delay at 3.55 minutes, followed by B6 at 1.77 minutes. In contrast, XY has the lowest average arrival delay at -16.16 minutes. Several airlines have negative average arrival delays, indicating that their flights arrived earlier than their scheduled arrival times on average within the available dataset.

## Key Findings

- The dataset contains 96,711 flight records and 35 columns.
- Only one missing value was found in the `fl_date` column.
- WN has the highest number of flights in the dataset.
- B6 has the highest average departure delay, while YX has the lowest.
- Day 6 has the highest average departure delay among the recorded days.
- Departure delay and arrival delay show a strong positive correlation of 0.73.
- Only 324 flights were cancelled, compared with 96,387 non-cancelled flights.
- The departure and arrival delay distributions show variation in flight timing performance.

```python
cancel_rate = (
    df.groupby("op_unique_carrier")["cancelled"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))
cancel_rate.plot(kind="bar")
plt.xlabel("Airline")
plt.ylabel("Cancellation Rate (%)")
plt.title("Cancellation Rate by Airline")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Observation

Cancellation rates vary across airlines. AS has the highest cancellation rate in the available dataset, followed by HA and B6. Several airlines have very low cancellation rates, while OH and YX show almost no cancellations. This indicates that cancellation frequency differs across airlines during the period covered by the dataset.

```python
# Cancellation status vs average departure delay

cancel_delay = df.groupby("cancelled")["dep_delay"].mean()

print("Average departure delay by cancellation status:")
print(cancel_delay.round(2))

plt.figure(figsize=(7, 5))
cancel_delay.plot(kind="bar")
plt.xlabel("Cancelled (0 = No, 1 = Yes)")
plt.ylabel("Average Departure Delay (minutes)")
plt.title("Average Departure Delay by Cancellation Status")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Correlation between delay causes and arrival delay

delay_causes = [
    "carrier_delay",
    "weather_delay",
    "nas_delay",
    "security_delay",
    "late_aircraft_delay"
]

cause_corr = df[delay_causes + ["arr_delay"]].corr()["arr_delay"].drop("arr_delay")

print("Correlation of delay causes with arrival delay:")
print(cause_corr.round(3))

plt.figure(figsize=(9, 5))
cause_corr.sort_values(ascending=False).plot(kind="bar")
plt.xlabel("Delay Cause")
plt.ylabel("Correlation with Arrival Delay")
plt.title("Delay Causes vs Arrival Delay")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Observation

The correlation between delay causes and arrival delay varies across the different causes. Late aircraft delay shows the strongest positive correlation with arrival delay (approximately 0.33), followed closely by NAS delay (approximately 0.32) and carrier delay (approximately 0.28). Weather delay and security delay show much weaker positive correlations, at approximately 0.07 and 0.06 respectively.

## Bivariate EDA Findings

- Departure delay and arrival delay show a strong positive correlation of **0.734**, indicating that longer departure delays generally correspond to longer arrival delays.
- Average arrival delay varies considerably across airlines, with HA showing the highest average and XY the lowest in the available dataset.
- Cancellation rates differ across airlines, with AS showing the highest cancellation rate during the recorded period.
- Among the examined delay causes, late aircraft delay, NAS delay, and carrier delay have stronger positive correlations with arrival delay than weather and security delays.
- These findings describe relationships in the available **January 1–6, 2024** dataset and should not be generalized to the entire year.
