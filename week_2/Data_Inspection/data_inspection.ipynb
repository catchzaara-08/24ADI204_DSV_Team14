## Airline Flight Delay Analysis

### Objective
To understand the structure, variables, data types, and quality of the flight delay dataset before performing data cleaning and exploratory data analysis.

### Dataset Details
- Dataset: Flight Data 2024
- Records: 3,538,856
- Features: 35
- Date range: January 1, 2024 – July 4, 2024

### Tools Used
- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn


```python

import pandas as pd
import numpy as np

file_path = r"C:\Users\ashun\Downloads\flight_data_2024_final_cleaned.csv"

print("Loading dataset...")

df = pd.read_csv(file_path, low_memory=False)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Display the first 5 records
display(df.head())

# Check column names
print("Column names:")
print(df.columns.tolist())

# Check dataset dimensions
print("\nDataset shape:", df.shape)

# Check data types and non-null counts
df.info()
```

```python
# Missing-value analysis
missing_count = df.isnull().sum()
missing_percent = (missing_count / len(df) * 100).round(2)

missing_report = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage (%)": missing_percent
})

missing_report = missing_report[missing_report["Missing Count"] > 0]
missing_report = missing_report.sort_values(
    by="Missing Count", ascending=False
)

display(missing_report)
```

```python
# Check for invalid flight status values
print("Cancelled values:")
print(df["cancelled"].value_counts(dropna=False))

print("\nDiverted values:")
print(df["diverted"].value_counts(dropna=False))

# Check whether flight dates are stored consistently
print("\nFlight date data type:")
print(df["fl_date"].dtype)

# Check for negative distances
print("\nNegative distances:", (df["distance"] < 0).sum())

# Check for negative scheduled flight durations
print(
    "Negative scheduled elapsed times:",
    (df["crs_elapsed_time"] < 0).sum()
)
```
